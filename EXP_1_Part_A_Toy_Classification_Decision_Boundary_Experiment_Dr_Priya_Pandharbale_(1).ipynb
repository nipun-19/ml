# Experiment: Toy Classification Using Decision Boundaries (Manual Coding)


## Aim
To implement a simple toy classification problem using a manually defined decision boundary and visualize the classification results.

## Step 1: Import Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Step 2: Create a Toy Dataset

The dataset has two input features (`X1`, `X2`) and two classes:
- Class 0
- Class 1

```python
X = np.array([
    [1, 1], [2, 1], [1, 2], [2, 2],
    [4, 4], [5, 4], [4, 5], [5, 5]
])

y = np.array([0, 0, 0, 0, 1, 1, 1, 1])

print("Input data:")
print(X)
print("Actual classes:")
print(y)
```

## Step 3: Define a Decision Boundary Manually

Use the line:

**X1 + X2 = 6**

- If `X1 + X2 >= 6`, predict Class 1.
- Otherwise, predict Class 0.

```python
y_pred = []

for point in X:
    if point[0] + point[1] >= 6:
        y_pred.append(1)
    else:
        y_pred.append(0)

print("Predicted classes:")
print(y_pred)
```

## Step 4: Calculate Accuracy

```python
correct = 0

for i in range(len(y)):
    if y[i] == y_pred[i]:
        correct += 1

accuracy = correct / len(y) * 100
print("Correct predictions:", correct)
print("Accuracy:", accuracy, "%")
```

## Step 5: Visualize the Dataset and Decision Boundary

```python
plt.scatter(X[:, 0], X[:, 1], c=y, edgecolors="black")

x_line = np.array([0, 6])
y_line = 6 - x_line

plt.plot(x_line, y_line, "r--", label="X1 + X2 = 6")
plt.xlabel("Feature X1")
plt.ylabel("Feature X2")
plt.title("Toy Classification with Manual Decision Boundary")
plt.legend()
plt.grid()
plt.show()
```

## Step 6: Test a New Data Point

Enter a new point and classify it using the same manually defined boundary.

```python
new_point = [3, 4]

if new_point[0] + new_point[1] >= 6:
    print("New point belongs to Class 1")
else:
    print("New point belongs to Class 0")
```

## Additional Example: Classify Students as Pass / Fail

The same idea with a different boundary. Students are described by **study hours** and **attendance (%)**. We define the rule `2 * hours + attendance / 10 >= 12` manually and test it.

```python
# Features: [Study hours per day, Attendance %]
students = np.array([
    [1, 40], [2, 50], [1, 60], [3, 45],
    [4, 80], [5, 85], [4, 90], [6, 95]
])
result = np.array([0, 0, 0, 0, 1, 1, 1, 1])   # 0 = Fail, 1 = Pass

def predict_student(hours, attendance):
    score = 2 * hours + attendance / 10
    return 1 if score >= 12 else 0

predicted = [predict_student(h, a) for h, a in students]
accuracy = np.mean(np.array(predicted) == result) * 100
print("Actual:   ", list(result))
print("Predicted:", predicted)
print("Accuracy:", accuracy, "%")

# Plot the boundary: 2*hours + attendance/10 = 12  ->  attendance = 10*(12 - 2*hours)
hours_line = np.linspace(0, 6, 50)
attendance_line = 10 * (12 - 2 * hours_line)

plt.scatter(students[:, 0], students[:, 1], c=result, edgecolors="black")
plt.plot(hours_line, attendance_line, "r--", label="2*hours + attendance/10 = 12")
plt.xlim(0, 7)
plt.ylim(30, 100)
plt.xlabel("Study Hours")
plt.ylabel("Attendance (%)")
plt.title("Pass / Fail Classification with a Manual Boundary")
plt.legend()
plt.grid()
plt.show()

print("New student (3 hours, 70% attendance):",
      "Pass" if predict_student(3, 70) == 1 else "Fail")
```

**Observation:** A manually chosen line can classify the data well, but it needs human judgement. Machine learning algorithms find such boundaries automatically from data.

## Observations

1. The toy dataset contains two features and two classes.
2. The decision boundary is defined manually using the equation `X1 + X2 = 6`.
3. Points with a sum greater than or equal to 6 are classified as Class 1.
4. Points with a sum less than 6 are classified as Class 0.
5. Accuracy is calculated by comparing actual and predicted class labels.
6. A new data point can also be classified using the same decision rule.
